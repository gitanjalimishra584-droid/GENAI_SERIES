### Prompt Templates - Dyanamic Prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
llm=ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
out= StrOutputParser()
```

```python
#### My self Chain Node

def transform_case(result:str):
    return result.upper()
```

```python
prompts= ChatPromptTemplate.from_messages([
    {"role":"system","content":"You are a translator and translate input in {language}"},
    {"role":"user","content":"{query}"},
]
)
prompts
```

```python
final_prompts= prompts.format_messages(language="hindi",query="what are u doing")
res=llm.invoke(final_prompts)
print(res.content)
```

### Chains- Runnable

```python
res=llm.invoke(prompts.invoke({"language":"Hindi","query":"I lovee You Suraj"}))
res.content
```

```python
chains= prompts| llm| out | transform_case
```

```python
res=chains.invoke({"language":"Hinglish","query":"Tum mere bubu hona"})
print(res)
```
